```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("/content/Country-data.csv")
df
```

```python
print("First 5 Records:")
print(df.head())
```

```python
X = df.drop('country',axis = 1)
X
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled[:5])
```

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
kmeans.fit(X_scaled)
```

```python
df['cluster'] = kmeans.labels_
print("\nCountries with Cluster:")
print(df[['country','cluster']].head(15))
```

```python
print("\nNumber of countries in each cluster:")
print(df['cluster'].value_counts().sort_index())
```

```python
new_sample = pd.DataFrame(
[[10, 40, 7, 35, 30000, 5, 75, 2, 15000]],
columns = X.columns
)
new_sample_scaled = pd.DataFrame( scaler.transform(new_sample), columns = new_sample.columns
)
new_clusters = kmeans.predict(new_sample_scaled)
```

```python
print("\n new country sample:")
print(new_sample)
print("\nCluster of new sample:")
print(new_clusters[0])
```
